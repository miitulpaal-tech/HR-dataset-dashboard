```python
columns_to_remove = ['Unnamed: 0','FirstName','LastName','ADEmail','DOB','State','Location','Title','Training Cost','Training Duration(Days)','Trainer','Location','Training Date','Survey Date','TerminationDescription','TerminationType','Supervisor']
```

```python
hr = hr.drop(columns=columns_to_remove)
```

```python
hr.info()
```

```python
hr.drop_duplicates()
```

```python
hr.isna().sum()
```

```python
hr['ExitDate'] = hr['ExitDate'].fillna("active")
```

```python
hr.head()
```

```python
hr.isna().sum()
```

```python
hr = hr.rename(columns={'RaceDesc':'Race'
                   ,'MaritalDesc':'Marital Status'})
```

```python
hr.info()
```

```python
hr['StartDate']= pd.to_datetime(hr['StartDate'])
```

```python
hr["EmployeeStatus"] = hr["EmployeeStatus"].replace({
    'Future Start':'Active',
    'Voluntarily Terminated':'Terminated',
    'Leave of Absence':'Active',
    'Terminated for Cause':'Terminated'
})
```
